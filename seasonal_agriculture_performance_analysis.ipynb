```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Find and load file
file_path = None
for root, dirs, files in os.walk('/Users/arc'):
    if 'seasonal_agriculture_performance_dataset.csv' in files:
        file_path = os.path.join(root, 'seasonal_agriculture_performance_dataset.csv')
        break

if file_path:
    df = pd.read_csv(file_path)
    
    # Create pivot table for heatmap
    crop_season_profit = df.pivot_table(
        values='Profit_INR', 
        index='Season', 
        columns='Crop', 
        aggfunc='mean'
    )
    
    # Plot heatmap
    plt.figure(figsize=(12,7))
    sns.heatmap(crop_season_profit, annot=True, fmt='.0f', cmap='RdYlGn', center=0)
    plt.title('Average Profit by Crop and Season')
    plt.xlabel('Crop')
    plt.ylabel('Season')
    plt.tight_layout()
    plt.show()
else:
    print("File not found. Drag CSV to project folder.")
```

# Key Findings

1. **Kharif season** recorded the highest average yield among the three seasons.

2. **Kharif** also achieved the highest average revenue and profit.

3. **Zaid season** showed the weakest economic performance, with negative average profit.

4. **Kharif** had the highest average water-use efficiency compared with Rabi and Zaid.

5. **Zaid** recorded the highest average water usage but the lowest water-use efficiency.

6. **Sugarcane** was the highest-yielding crop in the dataset.

7. **Sugarcane and Chilli** were among the most profitable crops.

8. **Wheat, Rice and Maize** showed comparatively lower average profitability.

9. **Drip irrigation** recorded the highest average yield and average profit among the irrigation methods.

10. **Kharif** had higher average rainfall, humidity and soil moisture than Rabi and Zaid.

11. **Zaid** was comparatively hotter and drier, with lower rainfall and soil moisture.

12. Agricultural performance varied considerably based on the **combination of season, crop, irrigation method and environmental conditions**.

# Conclusion

The Seasonal Agriculture Performance Analysis shows that agricultural outcomes vary considerably across seasons, crops, irrigation methods and environmental conditions.

Kharif emerged as the strongest-performing season, recording higher average yield, revenue, profit and water-use efficiency. In contrast, Zaid showed comparatively weaker economic performance, including negative average profit.

Crop-level analysis showed that Sugarcane and Chilli were among the most profitable crops, while Wheat, Rice and Maize experienced lower profitability in several seasonal combinations. Drip irrigation also demonstrated strong average yield and profit performance.

Overall, the analysis demonstrates that **seasonal conditions, crop selection, irrigation practices and environmental factors are important dimensions for understanding agricultural performance**. Data analytics and visualization can therefore help identify performance patterns and provide useful evidence for better agricultural planning and decision-making.
